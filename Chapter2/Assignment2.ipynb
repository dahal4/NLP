# Exercise 1: Word Frequency Counter
Objective : 
Count how many times each word appears in a sentence.

```python
input_text = """Three different types of unmanned weaponry. (Illustration by News Decoder)
This article was produced exclusively for News Decoder's global news service. It is through articles like this that News Decoder strives to provide context to complex global events and issues and teach global awareness through the lens of journalism. Learn how you can incorporate our resources and services into your classroom or educational program. 
“My main concern is with lethal autonomous weapons systems."""
```

```python
import re

# Convert to lowercase for case-insensitive counting
text = input_text.lower()

# Split into words (removing punctuation)
words = re.findall(r'\b\w+\b', text)

# Create dictionary to store word counts
word_counts = {}

for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)
```

# Exercise 2: Vocabulary Builder

Objective: 

Find all unique words in a corpus.

```python
import re

# Split input_text into sentences using regex to handle multiple sentence enders.
sentences = re.split(r'[.!?]\s*', input_text.strip())
vocab_set = set()

for sentence in sentences:
    # Extract words from each sentence
    words_in_sentence = re.findall(r'\b\w+\b', sentence.lower())
    vocab_set.update(words_in_sentence)

# Remove empty string if any (in case of trailing punctuation)
vocab_set.discard('')
print(vocab_set)

print("Vocabulary size:", len(vocab_set))
```

# Exercise 3: Build a Word Co-occurrence Matrix

Objective:

Construct a simple co-occurrence matrix using a window size of 1.

```python

input_sentence = input_text.split('\n')[0]  # "Three different types of unmanned weaponry."
# We'll take only the first sentence up to the first punctuation mark.
import re

first_sentence = re.split(r'[.!?]', input_sentence)[0].strip()

words = re.findall(r'\b\w+\b', first_sentence.lower())

# Build vocabulary (appearance order, unique)
vocab = []
for w in words:
    if w not in vocab:
        vocab.append(w)

# Initialize matrix
cooccurrence = {w: {w2: 0 for w2 in vocab} for w in vocab}
print(f"words: {words}")
print(f"vocab: {vocab}")
# Count neighboring words (window size=1)
for i, word in enumerate(words):
    neighbors = []
    if i > 0:
        neighbors.append(words[i-1])
    if i < len(words) - 1:
        neighbors.append(words[i+1])
    for n in neighbors:
        cooccurrence[word][n] += 1

# Print only the co-occurrence matrix without row/column labels
for w in vocab:
    print(" ".join(str(cooccurrence[w][w2]) for w2 in vocab))
```

# Exercise 4 : Detect Data Sparsity

Objective:

Find word pairs that never occur together.

```python
never_occurred = []
for w1 in vocab:
    for w2 in vocab:
        if w1 != w2 and cooccurrence[w1][w2] == 0:
            never_occurred.append((w1, w2))

unique_pairs = set()
for w1, w2 in never_occurred:
    if (w2, w1) not in unique_pairs:
        unique_pairs.add((w1, w2))
        print(f"({w1}, {w2}) -> Never occurred")
```

# Exercise 5: Implement Laplace Smoothing

Objective:

Calculate word probabilities using Laplace smoothing.

```python
# word_counts = {
#     "cat": 4,
#     "dog": 3,
#     "bird": 1
# }
# taking word count from the previous exercise
total_count = sum(word_counts.values())
vocab_size = len(word_counts)

smoothed_probs = {}
for word, count in word_counts.items():
    smoothed_prob = (count + 1) / (total_count + vocab_size)
    smoothed_probs[word] = smoothed_prob

for word, prob in smoothed_probs.items():
    print(f"{word} : {prob:.2f}")
```

# Exercise 6: Simple Keyboard Prediction

Objective:

Predict the most common next word.

```python
# Build the training corpus
corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP"
]

# Prepare bigram counts (dictionary of (current_word, next_word): count)
from collections import defaultdict, Counter

bigram_counts = defaultdict(Counter)

# Populate bigram counts from each sentence
for sentence in corpus:
    tokens = sentence.split()
    for i in range(len(tokens) - 1):
        curr_word = tokens[i]
        next_word = tokens[i + 1]
        bigram_counts[curr_word][next_word] += 1

# Input phrase (as specified)
input_phrase = "I like"
input_tokens = input_phrase.split()
last_word = input_tokens[-1]

# Next-word frequency suggestions
next_word_freqs = bigram_counts.get(last_word, {})

# Find top three predictions (sorted by frequency)
def prediction_sort_key(item):
    # item is a tuple (word, frequency)
    return (-item[1], item[0])

top_predictions = sorted(
    next_word_freqs.items(), 
    key=prediction_sort_key
)[:3]

# Output as requested
print("Suggestions")
for word, _ in top_predictions:
    print(word)
```

# Exercise 7: Calculate TF-IDF

Objective:

Compute TF, IDF, and TF-IDF scores.

```python
# Define corpus
corpus = [
    "I like NLP",      # Document 1
    "I like AI",       # Document 2
    "AI is amazing"    # Document 3
]

# Preprocess: tokenize and lowercase
tokenized_corpus = [doc.lower().split() for doc in corpus]

# Build vocabulary
vocabulary = sorted(set(word for doc in tokenized_corpus for word in doc))

# Calculate TF for each document
tf_list = []
for tokens in tokenized_corpus:
    tf = {}
    total_tokens = len(tokens)
    for word in vocabulary:
        tf[word] = tokens.count(word) / total_tokens
    tf_list.append(tf)

# Calculate DF for each word
from collections import Counter

df = {}
for word in vocabulary:
    df[word] = sum(1 for doc in tokenized_corpus if word in doc)

# Calculate IDF for each word
import math

N = len(tokenized_corpus)
idf = {}
for word in vocabulary:
    idf[word] = math.log(N / df[word]) if df[word] else 0.0

# Compute TF-IDF for each document
tfidf_list = []
for tf in tf_list:
    tfidf = {}
    for word in vocabulary:
        tfidf[word] = tf[word] * idf[word]
    tfidf_list.append(tfidf)

# Display results
print("Vocabulary:", vocabulary)
print("\nTF (Term Frequency) per document:")
for i, tf in enumerate(tf_list, 1):
    print(f"Doc {i}: {tf}")

print("\nDF (Document Frequency):")
print(df)

print("\nIDF (Inverse Document Frequency):")
print({word: round(score, 3) for word, score in idf.items()})

print("\nTF-IDF per document:")
for i, tfidf in enumerate(tfidf_list, 1):
    print(f"Doc {i}:")
    print({word: round(tfidf[word], 3) for word in vocabulary})
```
