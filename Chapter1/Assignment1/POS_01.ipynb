```python
import os
import requests
from bs4 import BeautifulSoup
import spacy
import pandas as pd
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
article_url = input("Enter the article URL: ")
txt_filename ="article.txt"
csv_filename = "ManishDahal_POS_01.csv"

#Extract content from url using beautifulsoup
response = requests.get(article_url)
if response.status_code == 200:
  soup = BeautifulSoup(response.content, 'html.parser')

  # taking only paragraphs ( get_text extracted all the info which is not relevent)
  paragraph = soup.find_all('p')
  with open(txt_filename, 'w', encoding='utf-8') as file:
    for p in paragraph:
      file.write(p.get_text() + '\n')
  print("Content extracted successfully")
else:
  print(f"Failed to retrieve content. Status code: {response.status_code}")
```

```python
with open(txt_filename, 'r', encoding='utf-8') as file:
  article_content = file.read()

document = nlp(article_content)
extracted_data = []
for token in document :
  if token.pos_ in ["NOUN","VERB"] and token.is_alpha:
    extracted_data.append({"Word":token.text,
                           "POS_Tag":token.pos_})
df_pos = pd.DataFrame(extracted_data)
print(df_pos)
df_pos.to_csv(csv_filename, index=False)
```

```python
extracted_ner=[]
for entity in document.ents:
  extracted_ner.append({"Entity":entity.text,
                        "Entity_type":entity.label_})
df_ner = pd.DataFrame(extracted_ner)
print(df_ner)
```
